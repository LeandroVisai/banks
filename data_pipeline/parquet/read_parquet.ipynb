```python
try:
    import pandas as pd
    # import succeeded — quick smoke check
    import numpy as np  # cell 2 already imports numpy, this is harmless if present
    print("pandas imported as pd, numpy version:", np.__version__, "pandas version:", pd.__version__)
except Exception as e:
    import sys, subprocess
    msg = str(e)
    print("Import error:", msg)
    if "numpy.dtype size changed" in msg or "binary incompatibility" in msg or "Expected" in msg:
        print("Reinstalling numpy, pandas and pyarrow (force-reinstall). The kernel will be restarted.")
        subprocess.check_call([sys.executable, "-m", "pip", "install", "--upgrade", "--force-reinstall", "--no-cache-dir", "numpy", "pandas", "pyarrow"])
        # restart the kernel to load the rebuilt C extensions
        from IPython import get_ipython
        get_ipython().kernel.do_shutdown(True)
    else:
        raise
```
